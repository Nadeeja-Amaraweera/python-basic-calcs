```python
# Extend list
my_list1 = ["dog","cat","snake","bird","fish"]
my_list1.extend(["lion","tiger","elephant"])
print(my_list1)
```

```python
# Remove specific item
my_list = ["dog","cat","snake","bird","fish"]
my_list.remove("snake")
print(my_list)
```

```python
# Clear the list
my_list = ["dog","cat","snake","bird","fish"]
my_list.clear()
print(my_list)
```

```python
# delete the list
numbers = [10,20,30,40,50]
del numbers[2]
print(numbers)
```

```python
my_list = ["dog","cat","snake","bird","fish"]
my_list.pop(2)
print(my_list)
```

```python
# reverse the list
my_list = ["dog","cat","snake","bird","fish"]
my_list.reverse()
print(my_list)
```

```python
# sort the list(Decending order)
my_list = ["dog","cat","snake","bird","fish"]
my_list.sort(reverse=True)
print(my_list)
```

```python
# sort the list(Ascending order)
my_list = ["dog","cat","snake","bird","fish"]
my_list.sort()
print(my_list)
```

```python
# extend the list
my_list1 = ["dog","cat","snake","bird","fish"]
my_list2 = ["lion","tiger","elephant"]
my_list1.extend(my_list2)
print(my_list1)
```

```python
# join the list
my_list1 = ["dog","cat","snake","bird","fish"]
my_list2 = ["lion","tiger","elephant"]
my_list3 = my_list1 + my_list2
print(my_list3)
```

```python
# if you want copy your previous list to new list you can use copy() method
my_list1 = ["dog","cat","snake","bird","fish"]
my_list2 = my_list1.copy()
print(my_list2)
```

```python
sales = [12000,15000,10000,16000,20000]
# the monthly sales of the company are folows,
# insert the 22000 at index 3
sales.insert(3, 22000)
# Sort the list in ascending order and calculate the percentage increase from the lowest sales value to the highest sales value
sales.sort()
print(sales)
percentage_increase = ((sales[-1] - sales[0])/sales[0])*100
print(f"The percentage increase from the lowest sales value to the highest sales value is: {percentage_increase}%")
```

```python
# A student's marks for five subject are as follows:
marks = [75, 82, 68, 91, 84]
# insert the marks of a new subject (88) at index 3
marks.insert(3,88)
# Calculate the total marks and print the result.
total_marks = sum(marks)
print(f"The total marks of the student is: {total_marks}")
# Calculate the percentage is 100%.
percentage_marks = (sum(marks) / (len(marks)*100))*100
print(f"The percentage marks of the student is: {percentage_marks}%")
```

```python
temperatures = [25.5, 30.2, 28.7, 32.1, 29.8]
#  insert 34 at index 2 , sort the list in ascending order and calculate the range
# range = max - min
temperatures.insert(2, 34)
temperatures.sort()
print(temperatures)
temperature_range = max(temperatures) - min(temperatures)
print(f"The range of the temperatures is: {temperature_range}°C")
```

```python
# tuples with only one item
my_tuple = ("apple",)
print(type(my_tuple))  # Output: <class 'tuple'>

my_tuple = ("apple")
print(type(my_tuple))  # Output: <class 'str'>
```

```python
# check negetive index
my_tuple = (1, 2, 3, 4, 5,)
print(my_tuple[2])  # Output: 3
# Checked it's OK
```

```python
my_tuple = ("tennis", 2, True, 4, 5.2,)
print(my_tuple[0:3])  # Output: ('tennis', 2, True)
```

```python
# Nested tuples
nested_tuple = ((1, 2), (3, 4), ("a", "b", "c"))
print(nested_tuple[0])  # Output: (1, 2)
print(nested_tuple[1][0])  # Output: 3
print(nested_tuple[2][1])  # Output: "b"
```

```python
# add values to a tuple
my_tuple = (1, 2, 3)
print(my_tuple)  # Output: <class 'tuple'>
list_1 = list(my_tuple)
list_1.append(4)
my_tuple = tuple(list_1)
print(my_tuple)  # Output: (1, 2, 3, 4
```
