```python
sales = [12000,15000,10000,16000,20000]
# the monthly sales of the company are folows,
# insert the 22000 at index 3
sales.insert(3, 22000)
# Sort the list in ascending order and calculate the percentage increase from the lowest sales value to the highest sales value
sales.sort()
print(sales)
percentage_increase = ((sales[-1] - sales[0])/sales[0])*100
print(f"The percentage increase from the lowest sales value to the highest sales value is: {percentage_increase}%")
```

```python
# A student's marks for five subject are as follows:
marks = [75, 82, 68, 91, 84]
# insert the marks of a new subject (88) at index 3
marks.insert(3,88)
# Calculate the total marks and print the result.
total_marks = sum(marks)
print(f"The total marks of the student is: {total_marks}")
# Calculate the percentage is 100%.
percentage_marks = (sum(marks) / (len(marks)*100))*100
print(f"The percentage marks of the student is: {percentage_marks}%")
```

```python
marks = [72,85,91,68,85,77,93,56,72,88,95,64,79,91,83]
print(marks)

# Sort the list in ascending order and print the result.
marks.sort()
print(marks)

# Sort the list in descending order and print the result.
marks.sort(reverse=True)
print(marks)

# Remove duplicates from the list and print the result.
marks = set(marks)
print(marks)

# Convert the unique marks back into a list and sort them in ascending order.
marks = list(set(marks))
marks.sort()
print(marks)

# Total marks
total_marks = sum(marks)
print(f"The total marks of the student is: {total_marks}")

# Average marks
average_marks = total_marks / len(marks)
print(f"The average marks of the student is: {average_marks}")

# Highest marks
highest_marks = max(marks)
print(f"The highest marks of the student is: {highest_marks}")

# Lowest marks
lowest_marks = min(marks)
print(f"The lowest marks of the student is: {lowest_marks}")

# Range of marks
range_of_marks = highest_marks - lowest_marks
print(f"The range of marks of the student is: {range_of_marks}")

# Calculate the variance each student's marks from the average marks and print the result.
variance = sum((x - average_marks) ** 2 for x in marks) / len(marks)
print(f"The variance of the marks is: {variance}")

# Calculate the standard deviation of the marks and print the result.
import math
standard_deviation = math.sqrt(variance)
print(f"The standard deviation of the marks is: {standard_deviation}")
```
