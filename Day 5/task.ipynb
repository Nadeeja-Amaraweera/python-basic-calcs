## The Shopping Cart Analyzer

```python
cart = [12.99, 45.50, 8.25, 150.00, 22.10, 99.99]
cart[2] = 9.50
print(cart)

cart.append(35.00)
print(cart)

cart.remove(150.00)
print(cart)

sum = 0
for price in cart:
    sum += price
    print(price)
print("Total:", sum)

print("----------------------")

new_cart = []
for price in cart:
    if price > 50:
        price *= 0.9
    new_cart.append(price)
print("Final cart:", new_cart)

print("----------------------")

budget_items = []
for price in new_cart:
    if price < 25:
        budget_items.append(price)
print("Budget items:", budget_items)
```

## Student Grade Processor

```python
grades = {"Alice": 85
          ,"Bob": 58
          , "Charlie": 92
          , "Diana": 45
          , "Evan": 78
          }
grades["Fiona"] = 88
print(grades)

print("----------------------")

for student, grade in grades.items():
    print("Student:", student, "\tGrade:", grade)

print("----------------------")

pass_count = 0
for grade in grades.values():
    if grade >= 60:
        pass_count += 1
print("Pass count:", pass_count)

print("----------------------")

max_grade = 0
max_student = ""
for student, grade in grades.items():
    if grade > max_grade:
        max_grade = grade
        max_student = student
print("Student Name: ", max_student, "\tMaximum grade:", max_grade)

print("----------------------")

new_point_students = {}
for student, grade in grades.items():
    if grade < 100:
        new_point_students[student] = grade + 5
    else:
        new_point_students[student] = grade
print("New point students:", new_point_students)
```

## Password Validation System

```python
password ="Py@th0n2023!"

if len(password) < 8:
    print("Too short.")
else:
    print("Valid length")

print("----------------------")

digit_count = 0
for char in password:
    if char.isdigit():
        digit_count += 1
print("Number of numeric digits in the password:", digit_count)

print("----------------------")

has_special = False
for char in password:
    if char in ['@', '#', '!']:
        has_special = True
        break
print("Has special character:", has_special)

print("----------------------")

print("First 3 characters:", password[0:3])
print("Last 4 characters:", password[-4:])

print("----------------------")

attempts = 3
while attempts > 0:
    attempts -= 1
    print(f"Attempt failed. {attempts} attempts remaining.")
```
