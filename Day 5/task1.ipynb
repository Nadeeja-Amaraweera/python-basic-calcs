## Find BMI

```python
weight = float(input("Enter your weight in kg: "))
height = float(input("Enter your height in meters: "))

def calculate_bmi(weight, height):
    bmi = weight / (height ** 2)
    print("Your BMI is:", bmi)

    if bmi < 18.5:
        category = "underweight"
    elif 18.5 <= bmi < 24.9:
        category = "normal"
    elif 25 <= bmi < 29.9:
        category = "overweight"
    else:
        category = "obese"

    return bmi, category

print(calculate_bmi(weight, height))
```

## Normalize Scores

The `normalize_scores` function accepts any number of scores and an optional `scaling_factor`. Each score is multiplied by the scaling factor. The default factor is `1.0`, so scores remain unchanged when no factor is provided.

```python
def normalize_scores(*scores, scaling_factor=1.0):
    """Scale any number of scores by the supplied factor."""
    return tuple(score * scaling_factor for score in scores)


print("Default factor:", normalize_scores(10, 20, 30))
print("Half scale:", normalize_scores(10, 20, 30, scaling_factor=0.5))
print("Double scale:", normalize_scores(1.5, 2.5, scaling_factor=2.0))
print("No scores:", normalize_scores(scaling_factor=3.0))
```

### Explanation of the test outputs

- `normalize_scores(10, 20, 30)` uses the default factor `1.0`, so the output is `(10.0, 20.0, 30.0)`.
- With `scaling_factor=0.5`, every score is halved, producing `(5.0, 10.0, 15.0)`.
- With `scaling_factor=2.0`, every score is doubled, producing `(3.0, 5.0)`.
- Calling the function without scores returns an empty tuple because there is nothing to scale.

The `*scores` parameter collects an arbitrary number of positional scores, while `scaling_factor` is keyword-only and defaults to `1.0`.

```python
def calculate_bill(customer,**items):
    sub_total = sum(items.values())
    if sub_total > 50:
        discount = sub_total * 0.1
    elif sub_total > 100:
        discount = sub_total * 0.2
    else:
        discount = 0
    amount_after_discount = sub_total - discount
    tax = amount_after_discount * 0.07
    print(f"Customer: {customer}")
    print(f"Sub Total: ${sub_total:.2f}")
    print(f"Discount: ${discount:.2f}")
    print(f"Tax: ${tax:.2f}")

    
calculate_bill("John Doe", Burger=5.99, Fries=2.49, Soda=1.50)
```
