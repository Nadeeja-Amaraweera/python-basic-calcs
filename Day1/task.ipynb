```python
list1 = [10, 2, 3, 4, 5]
print(list1[0])
print(list1[-1]) # Negative indexing starts from the end of the list
list1[0] = 100 # Modifying the first element of the list
print(list1)
```

```python
my_list = ["car", "bike", "bus", "train" , "plane" , "boat"]
print(my_list[0:3]) # Slicing the list from index 0 to 2
print(my_list[1:4]) # Slicing the list from index 1 to 3
print(my_list[2:]) # Slicing the list from index 2 to the end
print(my_list[:3]) # Slicing the list from the start to index 2
my_list[1:4] = ["scooter", "truck", "van"] # Modifying the list using slicing
print(my_list)
```
