```python
# Assending 
input1 = int(input("Enter first number: "))
input2 = int(input("Enter second number: "))
input3 = int(input("Enter third number: "))

if input1 <= input2 and input1 <= input3:
    if input2 <= input3:
        print(input1, input2, input3)
    else:
        print(input1, input3, input2)
elif input2 <= input1 and input2 <= input3:
    if input1 <= input3:
        print(input2, input1, input3)
    else:
        print(input2, input3, input1)
else:
    if input1 <= input2:
        print(input3, input1, input2)
    else:
        print(input3, input2, input1)
```

```python
# Structural pattern matching
response_code = 509
match response_code:
    case 200 | 201:
        print("OK")
    case 300 | 307:
        print("Redirect")
    case 400 | 401:
        print("Bad request")
    case 500 | 502:
        print("Internal Server Error")
    case _i:
        print("Unknown Error")
```

```python
my_list = [1,2]
match my_list:
    case []:
        print("The list is empty")
    case [x]:
        print("The list has one element", x)
    case [x, y]:
        print("The list has two elements", x, y)
    case [x, y, z]:
        print("The list has three elements", x, y, z)
    case _:
        print("The list has more than three elements")
```

```python
# Calculate the are of different jeomatrics shapes
shape_data = ["rectrangle", 2,5]

match shape_data:
    case ["circle", radius]:
        area = 3.14 * (radius**2)
        print("Circle Area: ",area)
    case ["square", side]:
        area = side ** 2
        print("Square area: ",area)
    case ["rectrangle", length,weight]:
        area = length * weight
        print("Reatrangle area: ",area)
    case _:
        print("Unknown Shape")
```

```python
# find factorial of a number
num = int(input("Enter a number: "))
factorial = 1
if num < 0:
    print("Factorial does not exist for negative numbers.")
else:
    for i in range(1, num + 1):
        print(i)
        factorial *= i
    print("The factorial of", num, "is", factorial)
```

```python
Inventory = {
    "Microchips":[50,12.50],
    "Resistors":[500,0.10],
    "Capacitors":[200,0.45]
}

total_inventory_value = 0
for item, details in Inventory.items():
    quantity = details[0]
    price = details[1]
    total_value = quantity * price
    print(f"{item}: Quantity = {quantity}, Price = {price}, Total Value = {total_value}")
    total_inventory_value += total_value
print("Total Inventory Value = ", total_inventory_value)
```
